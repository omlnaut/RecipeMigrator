```typescript
import { DefaultMealieClient, defaultRecipe } from "./_helper.ts";

const client = DefaultMealieClient()
const recipe = defaultRecipe;
```

```typescript
const slug = await client.UploadJsonRecipe(recipe);
```

```typescript
const filename = "_Baguette.jpg"
const fileBytes = await Deno.readFile(`/workspaces/RecipeMigrator/public/images/${filename}`)
const blob = new Blob([fileBytes], {type: 'image/jpg'})
```

```typescript
await client.UpdateImage(slug, {blob:blob, filename:filename})
```

```typescript
import { DefaultMealieClient, defaultRecipe } from "./_helper.ts";

const client = DefaultMealieClient()
const recipe = defaultRecipe;
```

```typescript
function GetEndpoint(slug: string): string {
  return `/api/recipes/${slug}/image`;
}
```

```typescript
const slug = await client.UploadJsonRecipe(recipe);
```

```typescript
const url = client.BuildUrl(GetEndpoint(slug));
url
```

```typescript
slug
```

```typescript
const fileBytes = await Deno.readFile("/workspaces/RecipeMigrator/public/images/_Baguette.jpg")
```

```typescript
const blob = new Blob([fileBytes], {type: 'image/jpg'})
```

```typescript
blob
```

```typescript
const formData = new FormData();
formData.append('image', blob, 'baguette.jpg');
formData.append('extension', 'jpg');
```

```typescript
const response = await fetch(url, {
      method: 'PUT',
      headers: {
        ... await client.AuthorizedDefaultHeader()
      },
      body: formData,
    });
```

```typescript
import { DefaultMealieClient } from "./_helper.ts";

const client = DefaultMealieClient()
```

```typescript
import type { ParsedRecipeProjection } from "../schemaOrgRecipe.ts";
import { ToSchemaOrgRecipe } from "../schemaOrgRecipe.ts";

const parsed: ParsedRecipeProjection = {
  title: "testi asdf",
  description: "testi mac testface",
  ingredients: ["first", "second"],
  instructions: ["do first", "then do second"],
};

const schemaOrg = ToSchemaOrgRecipe(parsed);
```

```typescript
await client.UploadJsonRecipe(parsed);
```

```typescript
const url = client.BuildUrl("/api/recipes/create/html-or-json");
const body = JSON.stringify({
  includeTags: false,
  data: JSON.stringify(schemaOrg),

  includeCategories: false,
  url: null,
});
const headers = {
  ...(await client.AuthorizedDefaultHeader()),
  "Content-Type": "application/json; charset=utf-8", // needed for deno(?)
};
```

```typescript
headers
```

```typescript
const preResponse = await fetch(url, {
  method: "POST",
  headers: headers,
  body: body,
});
```

```typescript
const slug = await preResponse.json()
```

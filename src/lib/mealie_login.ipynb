```typescript
const MEALIE_API_BASE = "http://192.168.2.116:9090"
```

```typescript
await fetch(MEALIE_API_BASE+"/docs")
```

```typescript
Deno.env.get("MEALIE_USERNAME")
```

```typescript
const body = new URLSearchParams({
  username: String(Deno.env.get("MEALIE_USERNAME") ?? ""),
  password: String(Deno.env.get("MEALIE_PASSWORD") ?? ""),
  remember_me: "false",
});
```

```typescript
const response = await fetch(MEALIE_API_BASE + "/api/auth/token", {
  method: "POST",
  headers: {
    "Content-Type": "application/x-www-form-urlencoded",
    accept: "application/json",
  },
  body,
})
```

```typescript
const raw_body = await response.json()
```

```typescript
console.assert("access_token" in raw_body)
```
